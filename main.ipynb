```python
import pandas as pd
import numpy as np
from gather_data import * 
from elo import *  
```

```python
# Gather Data
pas = gather_pas("2025-03-27", "2025-09-28")
```

```python
#Polish the PA Data
pas = pas.dropna(subset=['events'])
pas = pas[pas['events'] != "truncated_pa"] #pas cut short dont count towards the box score
pas.to_csv("pas.csv")
```

```python
pas["events"].unique()
```

```python
#Sort the games properly
pas = pas.sort_values(by=['game_date', 'game_pk', 'at_bat_number'], ascending=True)
pas.to_csv("pas.csv")
```

```python
#Begin building rankings

# these are going to hold batter and pitcher ratings
# these are just dicts
batters = {}
pitchers = {}

history = {}
```

```python
# fill in the batters

for i in pas['batter'].unique():
    batters[i.item()] = {"rating": 1000, "PAs": 0}

for i in pas['pitcher'].unique():
    pitchers[i.item()] = {"rating": 1000, "PAs": 0}
```

```python
# CALCULATE RATINGS

for row in pas.itertuples():
    batter = row.batter
    pitcher = row.pitcher
    
    b_before = batters[batter]['rating']
    p_before = pitchers[pitcher]['rating']

    b_after, p_after = elo(b_before, p_before, K)

    history.append({"date": row.game_date, 
                    "game": row.game_pk, 
                    "inning": row.inning,
                    "event": row.events,
                    "batter": batter,
                    "b_after": b_after,
                    "pitcher": pitcher,
                    "p_after": p_after})

    





```
